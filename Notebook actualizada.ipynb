```python
# Resumen: Localiza columnas con porcentajes y enseña valores de ejemplo
columnas_con_porcentaje = [columna for columna in df.columns if df[columna].dropna().astype(str).str.contains("%", regex=False).any()]  # Busca columnas con al menos un valor que contenga el símbolo de porcentaje
for columna in columnas_con_porcentaje:  # Recorre solo las columnas donde encontró porcentajes
    print(f"{columna}: {df[columna].dropna().astype(str).unique()[:5].tolist()}")  # Enseña hasta cinco valores de ejemplo de cada columna
```

```python
# Resumen: Crea df_clean y convierte seis columnas de porcentajes a proporciones
df_clean = df.copy()  # Crea una copia de df para conservar intacto el DataFrame original
columnas_porcentaje = ["Fuel Delivery - Diesel", "Fuel Delivery - Battery Electric Vehicle (BEV)", "Powertrain - Gasoline Mild Hybrid/MHEV", "Powertrain - Gasoline Strong Hybrid/HEV", "Powertrain - Gasoline with Start/Stop", "Powertrain - Gasoline without Start/Stop"]  # Indica las seis columnas cuyos valores incluyen porcentajes
for columna in columnas_porcentaje:  # Recorre cada columna de porcentajes
    df_clean[columna] = pd.to_numeric(df_clean[columna].str.rstrip("%"), errors="coerce") / 100  # Quita el símbolo (el strip elimina elementos del principio o del final a diferencia del replace)
df_clean[columnas_porcentaje].dtypes  # Comprueba que las seis columnas ahora tienen un tipo numérico
```

```python
# Resumen: Limpia y verifica todas las apariciones de Prelim. en Model Year
mask_prelim = df["Model Year"].str.contains("Prelim.", regex=False, na=False)  # Busca cualquier aparición literal de "Prelim." en la columna original
print(f"Filas con Prelim. antes de limpiar: {mask_prelim.sum()}")  # Cuenta todas las filas originales afectadas
print(df.loc[mask_prelim, "Model Year"].value_counts())  # Muestra todos los valores originales que contienen "Prelim." y sus frecuencias
df_clean["Model Year"] = df_clean["Model Year"].str.replace("Prelim.", "", regex=False).str.strip()  # Elimina todas las apariciones literales de "Prelim." y espacios sobrantes en la copia
prelim_restantes = df_clean["Model Year"].str.contains("Prelim.", regex=False, na=False).sum()  # Cuenta cualquier aparición que aún quede en toda la columna limpia
print(f"Apariciones de Prelim. restantes: {prelim_restantes}")  # Confirma que ya no queda ninguna
```

```python
# Resumen: Cuenta las celdas con valor All en cada columna
conteo_all = {}  # Prepara un diccionario para guardar el recuento por columna
for columna in df.columns:  # Recorre todas las columnas del DataFrame original
    valores = df[columna].astype("string").str.strip().str.casefold()  # Convierte los valores a texto, quita espacios y normaliza mayúsculas/minúsculas
    cuenta = valores.eq("all").fillna(False).sum()  # Cuenta celdas cuyo valor completo sea "All", ignorando los ausentes
    if cuenta > 0:  # Guarda solo las columnas donde encontró al menos un "All"
        conteo_all[columna] = cuenta  # Asocia el nombre de la columna con el número de filas encontradas
pd.Series(conteo_all, name="Filas con valor All").sort_values(ascending=False)  # Muestra las columnas con "All" ordenadas por frecuencia
```

```python
# Resumen: Filtra las filas agregadas con All y prepara df_modelo
columnas_all = ["Manufacturer", "Regulatory Class", "Vehicle Type"]  # Define las columnas donde encontramos valores All
mask_all = df_clean[columnas_all].astype("string").apply(lambda columna: columna.str.strip().str.casefold().eq("all").fillna(False)).any(axis=1)  # Marca cada fila que tiene All en una o más de esas columnas
print(f"Filas agregadas que se excluirían: {mask_all.sum()}")  # Cuenta filas únicas, aunque tengan All en varias columnas
df_modelo = df_clean.loc[~mask_all].copy()  # Crea una copia sin las filas agregadas para el análisis y el modelado
print(f"Filas que quedan: {len(df_modelo)}")  # Comprueba el tamaño del conjunto filtrado
```

```python
# Resumen: Interpreta las cuotas ausentes de Powertrain como cero para separar tecnologías
columnas_powertrain = [columna for columna in df_modelo.columns if columna.startswith("Powertrain - ")]  # Selecciona únicamente las columnas de cuotas Powertrain
ausentes_antes = int(df_modelo[columnas_powertrain].isna().sum().sum())  # Cuenta los valores ausentes en esas columnas antes del cambio
df_modelo[columnas_powertrain] = df_modelo[columnas_powertrain].fillna(0)  # Sustituye por cero las cuotas de tecnologías no presentes
ausentes_despues = int(df_modelo[columnas_powertrain].isna().sum().sum())  # Cuenta los valores ausentes restantes en Powertrain
print(f"Valores ausentes de Powertrain convertidos a 0: {ausentes_antes}; ausentes restantes: {ausentes_despues}")  # Resume el resultado de la conversión
df_modelo[columnas_powertrain].isna().sum()  # Verifica que cada columna Powertrain ya no tiene valores ausentes
```

```python
import seaborn as sns
import plotly.express as px
import matplotlib.pyplot as plt
```

```python
# Resumen: Comprueba si las cuotas Powertrain de cada fila suman aproximadamente 100 %
columnas_powertrain = [columna for columna in df_modelo.columns if columna.startswith("Powertrain - ")]  # Selecciona las columnas de cuotas Powertrain
powertrain_comprobacion = df_modelo[columnas_powertrain].copy()  # Crea una copia temporal para calcular las sumas sin cambiar df_modelo
for columna in columnas_powertrain:  # Recorre cada cuota Powertrain
    texto = powertrain_comprobacion[columna].astype("string").str.strip()  # Convierte temporalmente los valores a texto para detectar el símbolo %
    es_porcentaje = texto.str.endswith("%", na=False)  # Identifica los valores escritos como porcentajes
    valores = pd.to_numeric(texto.str.rstrip("%"), errors="coerce")  # Convierte los valores a números
    valores.loc[es_porcentaje] = valores.loc[es_porcentaje] / 100  # Convierte a proporción solo los valores que llevaban %
    powertrain_comprobacion[columna] = valores.fillna(0)  # Trata como cero los valores ausentes de estas cuotas
suma_powertrain = powertrain_comprobacion.sum(axis=1)  # Suma las cuotas Powertrain de cada fila
tolerancia = 0.01  # Acepta una diferencia máxima de un punto porcentual por redondeos
desviacion = suma_powertrain - 1  # Calcula cuánto se aleja cada fila de la suma esperada de 1 (100 %)
filas_cerca_100 = desviacion.abs() <= tolerancia  # Marca las filas cuya suma está entre 99 % y 101 %
print(f"Filas comprobadas: {len(suma_powertrain)}")  # Muestra cuántas filas se han revisado
print(f"Filas entre 99 % y 101 %: {filas_cerca_100.sum()} ({filas_cerca_100.mean() * 100:.2f} %)")  # Cuenta las filas que suman aproximadamente 100 %
print(f"Filas por debajo de 99 %: {(suma_powertrain < 1 - tolerancia).sum()}")  # Cuenta las filas cuya suma queda por debajo del rango esperado
print(f"Filas por encima de 101 %: {(suma_powertrain > 1 + tolerancia).sum()}")  # Cuenta las filas cuya suma supera el rango esperado
print(f"Suma mínima: {suma_powertrain.min() * 100:.2f} %; máxima: {suma_powertrain.max() * 100:.2f} %")  # Muestra los extremos de las sumas por fila
```

```python
# Resumen: Identifica posibles valores atípicos y muestra datos de los coches para revisarlos
q1 = df_modelo["Real-World MPG"].quantile(0.25)  # Calcula el primer cuartil del objetivo
q3 = df_modelo["Real-World MPG"].quantile(0.75)  # Calcula el tercer cuartil del objetivo
riq = q3 - q1  # Calcula el rango intercuartílico
limite_inferior = q1 - 1.5 * riq  # Calcula el límite inferior de la regla del boxplot
limite_superior = q3 + 1.5 * riq  # Calcula el límite superior de la regla del boxplot
print(f"Límites para revisar: {limite_inferior:.2f} a {limite_superior:.2f} MPG")  # Muestra los límites estadísticos
columnas_revisar = ["Manufacturer", "Model Year", "Vehicle Type", "Regulatory Class", "Real-World MPG", "Real-World MPG_City", "Real-World MPG_Hwy", "Weight (lbs)", "Engine Displacement", "Horsepower (HP)"]  # Elige datos que ayudan a contextualizar cada coche
posibles_atipicos = df_modelo[(df_modelo["Real-World MPG"] < limite_inferior) | (df_modelo["Real-World MPG"] > limite_superior)]  # Selecciona los valores fuera de los límites
posibles_atipicos[columnas_revisar].sort_values("Real-World MPG", ascending=False)  # Muestra primero los casos con mayor MPG
posibles_atipicos[columnas_revisar].sort_values("Real-World MPG", ascending=True)
```

```python
# Resumen: Crea un conjunto sin cuotas de híbridos ni vehículos eléctricos
categorias_electrificadas = [  # Lista las tecnologías cuyas filas queremos excluir si tienen una cuota mayor que cero
    "Powertrain - Gasoline Mild Hybrid/MHEV",  # Incluye los híbridos ligeros
    "Powertrain - Gasoline Strong Hybrid/HEV",  # Incluye los híbridos completos
    "Powertrain - Battery Electric Vehicle (BEV)",  # Incluye los eléctricos de batería
    "Powertrain - Plug-in Hybrid Electric Vehicle (PHEV)",  # Incluye los híbridos enchufables
    "Powertrain - Fuel Cell Electric Vehicle (FCEV)"  # Incluye los eléctricos de pila de combustible
]  # Termina la lista de categorías electrificadas
mask_noEV = df_modelo[categorias_electrificadas].eq(0).all(axis=1)  # Conserva solo las filas con cuota cero en todas esas categorías
df_modelo_noEV = df_modelo.loc[mask_noEV].copy()  # Crea el nuevo DataFrame sin modificar df_modelo
print(f"Filas en df_modelo original: {len(df_modelo)}")  # Muestra el tamaño del conjunto original
print(f"Filas sin cuota híbrida ni eléctrica: {len(df_modelo_noEV)}")  # Muestra el tamaño del conjunto filtrado
print(f"Filas excluidas: {len(df_modelo) - len(df_modelo_noEV)}")  # Cuenta las filas excluidas por tener alguna cuota electrificada
```

```python
# Resumen: Dibuja un boxplot estático y resume los cuartiles de Real-World MPG sin híbridos ni eléctricos
mpg_noEV = df_modelo_noEV["Real-World MPG"].dropna()  # Selecciona el objetivo y descarta posibles ausentes
q1_noEV = mpg_noEV.quantile(0.25)  # Calcula el primer cuartil (25 %)
mediana_noEV = mpg_noEV.quantile(0.50)  # Calcula la mediana (50 %)
q3_noEV = mpg_noEV.quantile(0.75)  # Calcula el tercer cuartil (75 %)
riq_noEV = q3_noEV - q1_noEV  # Calcula el rango intercuartílico
limite_inferior_noEV = q1_noEV - 1.5 * riq_noEV  # Calcula el límite inferior para señalar posibles atípicos
limite_superior_noEV = q3_noEV + 1.5 * riq_noEV  # Calcula el límite superior para señalar posibles atípicos
valores_no_atipicos = mpg_noEV[(mpg_noEV >= limite_inferior_noEV) & (mpg_noEV <= limite_superior_noEV)]  # Selecciona valores dentro de los límites estadísticos
print(f"Número de observaciones: {len(mpg_noEV)}")  # Muestra cuántos valores de MPG se analizaron
print(f"Mínimo: {mpg_noEV.min():.2f} MPG")  # Muestra el mínimo observado
print(f"Q1 (25 %): {q1_noEV:.2f} MPG")  # Muestra el primer cuartil
print(f"Mediana (Q2, 50 %): {mediana_noEV:.2f} MPG")  # Muestra la mediana
print(f"Q3 (75 %): {q3_noEV:.2f} MPG")  # Muestra el tercer cuartil
print(f"Máximo: {mpg_noEV.max():.2f} MPG")  # Muestra el máximo observado
print(f"Rango intercuartílico (Q3 - Q1): {riq_noEV:.2f} MPG")  # Muestra la amplitud del 50 % central de los datos
print(f"Límites estadísticos para posibles atípicos: {limite_inferior_noEV:.2f} a {limite_superior_noEV:.2f} MPG")  # Muestra los límites de 1,5 veces el rango intercuartílico
print(f"Bigotes del boxplot: {valores_no_atipicos.min():.2f} a {valores_no_atipicos.max():.2f} MPG")  # Muestra los valores observados extremos dentro de los límites
print(f"Posibles valores atípicos: {(~mpg_noEV.between(limite_inferior_noEV, limite_superior_noEV)).sum()}")  # Cuenta observaciones fuera de los límites, sin asumir que sean errores
plt.figure(figsize=(9, 3))  # Define el tamaño del gráfico estático
sns.boxplot(x=mpg_noEV, color="skyblue")  # Dibuja el boxplot estático de MPG
plt.title("Real-World MPG sin híbridos ni eléctricos")  # Añade un título descriptivo
plt.xlabel("Real-World MPG")  # Etiqueta el eje horizontal
plt.tight_layout()  # Ajusta los márgenes del gráfico
plt.show()  # Muestra el gráfico estático
```

```python
# Resumen: Compara el MPG desde 2010 según la presencia de gasolina con Start/Stop
columna_start_stop = "Powertrain - Gasoline with Start/Stop"  # Guarda el nombre de la columna Start/Stop
anios_modelo = pd.to_numeric(df_modelo_noEV["Model Year"], errors="coerce")  # Convierte el año del modelo a número
filas_desde_2010 = anios_modelo >= 2010  # Selecciona coches del año 2010 en adelante
filas_con_dato = filas_desde_2010 & df_modelo_noEV[columna_start_stop].notna()  # Excluye filas sin dato de año o Start/Stop
mpg_sin_start_stop = df_modelo_noEV.loc[filas_con_dato & (df_modelo_noEV[columna_start_stop] == 0), "Real-World MPG"].dropna()  # Obtiene el MPG de coches sin Start/Stop
autos_con_start_stop = filas_con_dato & (df_modelo_noEV[columna_start_stop] > 0)  # Identifica coches con cuota Start/Stop positiva
mpg_con_start_stop = df_modelo_noEV.loc[autos_con_start_stop, "Real-World MPG"].dropna()  # Obtiene el MPG de coches con Start/Stop
plt.boxplot([mpg_sin_start_stop, mpg_con_start_stop], tick_labels=["Sin Start/Stop", "Con Start/Stop"])  # Dibuja las cajas con sus categorías en el eje X
plt.ylabel("Real-World MPG")  # Etiqueta el eje vertical con el MPG
plt.title("MPG desde 2010 según Start/Stop")  # Añade un título que indica el filtro aplicado
plt.show()  # Muestra el gráfico
```
